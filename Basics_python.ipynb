```python
## Numeric data types
## 1. Integers
x = 5
y = -15
print(x,y)
```

```python
## floating numbers
x = 11.5
y = -7.5
print(x,y)
```

```python
## Complex numbers
x = 3 + 7j
y = 8 + 9j
print(x,y)
```

```python
## Boolean numbers
x = False
y = True
print(x,y)
```

```python
## Sequence data type
x = "hello"
y = "world"
print(x,y)
```

```python
# Lists
x = ['one', 2, 3, 'four', 5]
y = [ 'Mango', 'Kivi' 'Banana', 'Apple']
print(x,y)
```

```python
## Tuples
x = ('one', 2, 3, 'four', 5)
y = ( 'Mango', 'Kivi' 'Banana', 'Apple')
print(x,y)
```

```python
## Range objects
x = range(0,20)
for i in x:
    print(i)
```

```python
## Mapping data types 
## dictionaries
x = {'name':'John', 'age': 25 , 'city': 'Pune'}
y = { 1: 'one', 2: 'two', 3: 'three'}
print(x,y)
```

```python
## Set Data Types
# Sets
x = {1,2,3,4}
y = {5,6,7,8}
print(x,y)
```

```python
## Exploring data types
## For Loop
fruits = ['Mango', 'Apple', 'Kivi', 'Banana']
for fruit in fruits:
    print(fruit)
```

```python
# Looping through a string
name = 'Varsha'
for character in name:
    print(character)
```

```python
## Loopinf through the Dic
person = {'name': 'Rupa', 'age':27}
for key, value in person.items():
    print(key,value)
```

```python
## while loop
count = 0
while count < 15:
    print(count)
    count +=1
```

```python
## looping until a user a valid input
valid_input = False
while not valid_input:
    user_input = input(" Enter a number:")
    if user_input.isdigit():
        print('You entered:', user_input)
        valid_input = True
    else:
        print("Invalid input, please try again")
```

```python
# Looping until users decide to quit
while True:
    user_input = input('Enter a number or type "quit" to exit: ')
    if user_input == 'quit':
        break
    elif user_input.isdigit():
        print('You entered:', user_input)
    else:
        print('Invalid input, please try again')
```

```python
## functional variable length keyword argument
def print_kwargs(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} = {value}")
print_kwargs(name='Varsha', age=26)
```

```python
##A recursive function that calculates the factorial of a number
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)

result = factorial(8)
print(result)
```

```python
## Global & Local variable
def square(x):
    result = x ** 2
    return result
print(square(7))
    
```
